```python
import json, pandas as pd
grid = pd.read_parquet('var/denue/analysis/grid_features.parquet')
lq = pd.read_parquet('var/denue/analysis/grid_sector_lq.parquet')
maz = grid.query("municipality_code == '25_012'").copy()
display(maz.groupby(['source_edition','edition_date'], as_index=False).agg(establishments=('establishments','sum'), active_grids=('grid', lambda s: (maz.loc[s.index,'establishments'] > 0).sum()), mean_effective_sectors=('effective_sectors','mean')).tail(10))
display(lq.query("municipality_code == '25_012' and source_edition == '2026-05' and lq_municipality >= 1.25").sort_values('lq_municipality', ascending=False).head(30))
```
